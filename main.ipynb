```python
## 1. Importação e processamento dos produtos

from funcoes import *

produtos = ler_csv("olist_products_dataset.csv")

produtos, nulos_corrigidos = tratar_nulos(produtos)

produtos = padronizar_categorias(produtos)

produtos, dimensoes_corrigidas = tratar_dimensoes_nulas(produtos)

produtos, descritivos_corrigidos = tratar_colunas_descritivas(produtos)
```

```python
## 2. Processamento dos pedidos

pedidos = ler_csv("olist_orders_dataset.csv")

colunas_para_tratar = [
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "order_delivered_carrier_date",
    "order_purchase_timestamp",
]

total_datas_convertidas = 0

for coluna in colunas_para_tratar:
    datas_convertidas = converter_data(pedidos, coluna)
    total_datas_convertidas += datas_convertidas

cancelados, inconsistentes = conferir_status_pedidos(pedidos)
```

```python
## 3. Exportação dos dados tratados

salvar_csv(
    produtos,
    "olist_products_dataset_tratado.csv"
)

salvar_csv(
    pedidos,
    "olist_orders_dataset_tratado.csv"
)
```

```python
## 4. Relatório final 
## É exibido um resumo estatístico do processamento, incluindo total de registros, nulos corrigidos, datas convertidas e pedidos inconsistentes identificados.

print("===== RELATÓRIO FINAL =====")

print(f"Total de produtos processados: {len(produtos)}")

print(f"Total de pedidos processados: {len(pedidos)}")

print(f"Nulos categóricos corrigidos: {nulos_corrigidos}")

print(f"Dimensões físicas corrigidas: {dimensoes_corrigidas}")

print(f"Campos descritivos corrigidos: {descritivos_corrigidos}")

print(f"Total de datas convertidas: {total_datas_convertidas}")

print(f"Pedidos cancelados identificados: {cancelados}")

print(f"Pedidos inconsistentes: {inconsistentes}")
```
